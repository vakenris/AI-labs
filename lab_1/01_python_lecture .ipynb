## 4. Числа, строки и индексы

### Что мы сделаем?

Посмотрим арифметические операции и извлечём отдельные символы из строки.

### Зачем?

Индексация и срезы нужны для обработки текстовых данных, имён файлов, ответов пользователя и признаков, представленных строками.

```python
x = 10
y = 3
print('Сложение:', x + y)
print('Деление:', x / y)
print('Целая часть:', x // y)
print('Остаток:', x % y)
print('Степень:', x ** 2)

name = 'Python'

# Первый символ строки — первый индекс
print(name[0])        # P

# Последний символ строки — отрицательный индекс
print(name[-1])       # n

# Срез строки: символы с 1 по 3 (не включая 4)
print(name[1:4])      # yth

# Нижний регистр — все буквы маленькие
print(name.lower())   # python

# Замена символа 'P' на 'J'
print(name.replace('P', 'J'))   # Jython

assert name[0] == 'P'
assert name[1:4] == 'yth'
```

Индекс первого символа равен 0, а -1 означает последний символ. Срез name[1:4] берёт индексы 1, 2 и 3, но не включает правую границу 4.

## 5. Основные коллекции

### Что мы сделаем?

Создадим list, tuple, set и dict, а затем изменим список и словарь.

Создадим список, кортеж, множество и словарь, а затем изменим список и словарь.
### Зачем?

Коллекции позволяют хранить несколько объектов. Выбор структуры данных влияет на то, как легко добавлять элементы, искать значения и устранять повторы.

```python
numbers = [10, 20, 20, 30]     # список (list) — изменяемый, упорядоченный набор


coordinates = (55.75, 37.62)   # кортеж (tuple) — неизменяемый, упорядоченный набор


unique_numbers = set(numbers)  # множество (set) — уникальные, неупорядоченные элементы


student = {'name': 'Ирина', 'score': 20}  # словарь (dict) — пары «ключ: значение»

numbers.append(40)                          # добавляем элемент в список
student['passed'] = True                    # добавляем пару в словарь
student['score'] = 25                       # изменяем значение в словаре

print(numbers)                              # выводим список
print(coordinates[0])                       # выводим первый элемент кортежа
print(unique_numbers)                       # выводим множество
print(student['name'], student['score'])    # выводим значения из словаря

assert len(unique_numbers) == 3             # проверяем: в множестве 3 элемента
assert student['passed'] is True            # проверяем: passed равно True
```

list сохраняет порядок и изменяется; tuple обычно используют для фиксированной группы значений; set оставляет только уникальные элементы; dict связывает ключ со значением.

## 6. Условия, циклы и enumerate

### Что мы сделаем?

Выберем сообщение по оценке, сложим числа циклом и выведем индекс каждого элемента.

### Зачем?

Условие описывает выбор, цикл — повторяющееся действие. Это основа почти любой обработки набора данных.

### Наглядная схема

Сначала вычисляется условие. Выполняется только одна ветвь: if, если результат True, или else, если результат False.

## 7. Функции: превращаем решение в инструмент

### Что мы сделаем?

Напишем две функции из задания: разность чисел и функцию с несколькими ветвями.

### Зачем?

Функция даёт имя решению и позволяет использовать его много раз. return возвращает результат вызывающему коду; print только показывает результат на экране.

### Наглядная схема

В определении функции имена first, second или number_1 — это параметры, то есть локальные имена для входных значений. При вызове функции реальные значения становятся аргументами и передаются по позиции.
function_arguments_preview.png

### Параметры и аргументы: идея

В записи:

~~~python
def difference_of_two_numbers(first, second):
    return first - second
~~~

first и second — не специальные слова Python. Это имена параметров, которые выбирает автор функции. Можно написать a и b, смысл не изменится.

При вызове:

~~~python
difference_of_two_numbers(10, 4)
~~~

Python временно создаёт соответствие:

- first = 10;
- second = 4.

После этого выполняется тело функции, а return возвращает результат вызывающему коду. Значения в вызове называются аргументами. Важно различать определение функции и вызов функции.

```python
def difference_of_two_numbers(first, second):
    return first - second                            # возвращаем разность

# проверяем и печатаем результаты
result1 = difference_of_two_numbers(10, 4)
print(result1)                                       # выводим: 6
assert result1 == 6                                  # проверяем: равно 6
```

```python

def condition_function(input_number):
    if input_number <= 0:
        return input_number * 2                      # если <= 0, умножаем на 2
    if input_number <= 10:
        return input_number * 3                      # если <= 10, умножаем на 3
    return input_number / 10                         # иначе делим на 10

result2 = condition_function(-1)
print(result2)                                       # выводим: -2
assert result2 == -2                                 # проверяем: равно -2

result3 = condition_function(5)
print(result3)                                       # выводим: 15
assert result3 == 15                                 # проверяем: равно 15

result4 = condition_function(20)
print(result4)                                       # выводим: 2.0
assert result4 == 2                                  # проверяем: равно 2
```

В condition_function порядок проверок важен: сначала обрабатываются числа до нуля, затем числа до 10, а всё остальное попадает в последнюю ветвь.

## 8. Прикладные функции: калькулятор и подсчёт

### Что мы сделаем?

Соединим условие, арифметику, список и словарь в нескольких небольших функциях.

### Зачем?

Это уже похожий на реальные данные сценарий: принять вход, обработать его и вернуть проверяемый результат. В тетради студента эти функции нужно будет написать самостоятельно.

```python
def calculator(number_1, operation, number_2):
    if operation == '+':
        return number_1 + number_2                # сложение
    if operation == '-':
        return number_1 - number_2                # вычитание
    if operation == '*':
        return number_1 * number_2                # умножение
    if operation == '/':
        return number_1 / number_2                # деление
    raise ValueError('Неизвестная операция')      # неизвестная операция

# проверяем и печатаем результаты
r1 = calculator(4, '*', 3)
print(r1)                                         # выводим: 12
assert r1 == 12                                   # проверяем: равно 12

r2 = calculator(2, '/', 2)
print(r2)                                         # выводим: 1.0
assert r2 == 1                                    # проверяем: равно 1
```

```python

def number_of_unique_elements(input_list):
    return len(set(input_list))                   # количество уникальных


def counter(input_list):
    result = {}                                   # пустой словарь
    for element in input_list:
        result[element] = result.get(element, 0) + 1   # увеличиваем счётчик
    return result

r3 = number_of_unique_elements([1, 2, 1, 3])
print(r3)                                         # выводим: 3
assert r3 == 3                                    # проверяем: равно 3

r4 = counter([1, 1, 2, 3])
print(r4)                                         # выводим: {1: 2, 2: 1, 3: 1}
assert r4 == {1: 2, 2: 1, 3: 1}                   # проверяем словарь

print('Прикладные функции работают')              # всё работает
```

В counter метод get возвращает текущее количество или 0, если ключ встречается впервые. Важно не использовать готовый collections.Counter: цель упражнения — потренировать словари и цикл.

## 9. Строки, время и математическая формула

### Что мы сделаем?

Разберём строку с числами, преобразуем минуты в часы и напишем функции с библиотекой math.

### Зачем?

Здесь встречаются типичные шаги обработки: split, преобразование типа, цикл, остаток от деления и аккуратное округление результата.

```python
import math


def multiply_nums(input_string):
    result = 1
    for part in input_string.split(','):
        result *= int(part.strip())
    return result


def custom_function(x):
    return math.sin(x) * math.cos(x)


def time_converter(minutes):
    minutes_in_day = 24 * 60
    minutes = minutes % minutes_in_day
    hours, remainder = divmod(minutes, 60)
    return f'{hours} {remainder}'


assert multiply_nums('2, 3') == 6
assert round(custom_function(1), 3) == 0.455
assert time_converter(2782) == '22 22'
print('Строковая и математическая обработка работает')
```

split(',') разбивает строку, strip() убирает пробелы, а int() превращает текст в число. Для времени удобно сначала оставить минуты в пределах суток с помощью %.

Во второй математической задаче из лабораторной используется произведение по индексу k от 1 до n. Переменная цикла должна иметь другое имя, чтобы не путать текущий множитель с верхней границей.

```python
def custom_function_1(x, n):
    result = 1
    for k in range(1, n + 1):
        factor = ((k + 2) ** x + math.log(x)) / (x ** 2 + 4 * k)
        result *= factor
    return result


assert round(custom_function_1(2, 3), 3) == 2.707
assert round(custom_function_1(3, 2), 3) == 8.277
print('Произведение рассчитано')
```

```python
def min_of_three_values(first, second, third):
    minimum = first
    if second < minimum:
        minimum = second
    if third < minimum:
        minimum = third
    return minimum


def find_max(values):
    max_value = values[0]
    max_index = 0
    for index, value in enumerate(values):
        if value > max_value:
            max_value = value
            max_index = index
    return max_value, max_index


assert min_of_three_values(18, 7, 11) == 7
assert find_max([69, 16, 64, 54, 36, 70, 89, 29]) == (89, 6)
print('Поиск минимума и максимума работает')
```

В find_max мы храним не только максимальное значение, но и его индекс. Поэтому при обновлении максимума нужно обновить обе переменные.

## 11. Классы и наследование

### Что мы сделаем?

Создадим объект транспортного средства с атрибутами max_speed и mileage, а затем унаследуем класс Truck.

### Зачем?

Класс объединяет данные и операции над ними. На первом занятии достаточно понять конструктор __init__ и идею наследования; сложные архитектуры пока не нужны.

### Наглядная схема

При Vehicle(100, 50) значения в скобках передаются в __init__. self Python передаёт автоматически: это ссылка на создаваемый объект.

Передача аргументов конструктору класса


### Что означают значения в скобках у класса?

В определении конструктора:

~~~python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
~~~

max_speed и mileage — параметры конструктора. При создании объекта:

~~~python
vehicle = Vehicle(100, 50)
~~~

100 попадает в max_speed, а 50 — в mileage. Так объект получает собственные значения атрибутов.

Параметр self означает «этот конкретный объект». При обычном вызове Vehicle(100, 50) студент не передаёт self вручную.

Smartiqa_Python_Obje.png

```python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage


class Truck(Vehicle):
    pass


vehicle = Vehicle(100, 50)
truck = Truck(80, 1200)

assert (vehicle.max_speed, vehicle.mileage) == (100, 50)
assert (truck.max_speed, truck.mileage) == (80, 1200)
print(vehicle.max_speed, truck.mileage)
```

self — это конкретный объект. В Truck(Vehicle) класс-наследник получает конструктор родительского класса, если мы не переопределяем его.
